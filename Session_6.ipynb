**if-elif-else**

- When there is one condition is there
   
   - We can expect two answers
        
        - in that first answer under **if** Block
        
        - Second Answer is under **else** Block

- When there are two conditions

  - We can expect three answer
  
    - in that first answer under **if** block based on condition-1
    - Second answer under **elif** block based on conditioned - 2
    - Third anser under **else** block

```python
one condition
```

```python
if <condition>:
    <write your code>

else <condition>:
     <write your code>

else:
    <write your code>
    
```

```python
# wap ask the user enter a number 
# if the number equal to 1 print one
#otherwise if the equal to 2 then print two
#otherwise if the equal to 3 then print three
#otherwise print enter a valid number

num=eval(input('enter a number:'))
if num == 1:
    print('one')
elif num == 2:
    print('two')
elif num == 3:
    print('three')
else:
    print('enter a valid number')
```

```python
# wap ask the user enter percentage
# if percentage greator the 75 print A grade
# elif percentage between 60 to 75 print B grade
# elif percentage between 45 to 60 print C grade
# elif percentage between 30 to 45 print D grade
# otherwise print fail 

p = eval( input ("Enter The Percentage:") )
if p>75:
    print("A Grade")
elif p>60:
    print("B Grade")
elif p>45:
    print("C Grade")
elif p>30:
    print("D Grade")    
else:
    print("Fail")
```
