- Whenver we want to use interger then we are playing **int**
- Similarly if we want to use decimal number we are playing **float**
- There is one more operation is there: **eval**
- **eval** will convert integer values to int type,decimal values to float type

```python
int('20')
```

```python
float('30.5')
```

```python
# average of 3 numbers

n1=eval(input('enter the number1:'))
n2=eval(input('enter the number2:'))
n3=eval(input('enter the number3:'))
avg=(n1+n2+n3)/3
print('the average of n1, n2, n3 is:',avg)              
```

```python
# wap ask the user take two number print substraction multiplication and division using eval as input

n1=eval(input('enter the number1:'))
n2=eval(input('enter the number2:'))
subtraction=n1-n2
multiplication=n1*n2
division=n1/n2
print(f'The Multiplication of n1,n2 is :{multiplication},The Substraction of n1,n2 is :{subtraction},the Division of n1,n2 is :{division},')
```

```python
# wap ask the user take the radius find the area of circle
# pi=3.14
# r=100
# area_of_circle=pi*r*r

pi=3.14
r=eval(input('Enter the Radius'))
area_of_circle=pi*r*r
print(f'the area of the circle is:{area_of_circle},')
```

```python
# wap ask the user take height and breadth find the area of traingle
# area_of_traingle=0.5*b*h

height=eval(input('Enter the Height of the Traingle'))
breadth=eval(input('Enter the Breadth of the Traingle'))
area_of_traingle=0.5*breadth*height
print(f'the area of the traingle is:={area_of_traingle}')
```
