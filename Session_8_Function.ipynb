**args-kwargs**

- args means we can provide multiple value with out Argument name

    - which means while calling fuction we can provide driectly 'virat',102  

    - while define the fuction you need to mention argument names also

    - the output will comes in **tuple** format : ('virat',102)
    
- kwargs keyword arguments means we can provide multiple values with argument name

     - while define the function no need to provide the name of the arguments

     - while call the function definetly we need to provide with argument names

         - name='virat',runs=102

     - the output comes in dict format: {'name': 'virat','runs':102}

```python
#
#     print (b)
#     
def fun1(*args):
    print(args)
fun1(10,20)    
# with out name
# output in tuple format
```

```python
def fun2(**kwargs):
    print(kwargs)      # {a':10, 'b' :20}
fun2(a=10,b=20)
# with name
# output in dict format
```
