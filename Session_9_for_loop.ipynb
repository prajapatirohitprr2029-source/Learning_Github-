**case-3: range( start, stop, step)**

- start value = start

- if step is positive sign

    - it is positive direction
 
    - increment by **step value**
 
    - last value = stop-1
 
- if step is negative sign

   - it is a negative direction(decrement)
 
   - decerment by **step value**
 
   - last value = stop+1
